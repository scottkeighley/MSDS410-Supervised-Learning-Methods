#### Question 2

**Write out the null and alternate hypotheses for the t-test for Beta1.**

Test of whether $X_1$ has a significant linear relationship with the response, holding the other predictors constant: 
$$H_0 = \beta_1 = 0$$
$$H_a : \beta_1 \neq 0$$

#### Question 3

**Compute the t-statistic for Beta1. Conduct the hypothesis test and interpret the result.**

$$t = \frac{\hat \beta_1 - 0}{SE(\hat \beta_1)}$$

with $df = n - k - 1$, where $k$ = number of predictors in Model 1 (4). 

```python
est_X1 = model1['coef']['X1']['est']
se_X1 = model1['coef']['X1']['se']
k1 = model1['n_predictors']
df_resid1 = n - k1 - 1

t_X1 = (est_X1 - 0) / se_X1
p_X1 = 2 * (1 - stats.t.cdf(abs(t_X1), df_resid1))
t_crit = stats.t.ppf(0.975, df_resid1)

print(f"df (residual) = {df_resid1} (matches ANOVA Residuals df = {model1['anova']['Residuals']['df']})")
print(f"t = {est_X1} / {se_X1} = {t_X1:.4f}")
print(f"two-sided p-value = {p_X1:.6f}")
print(f"critical t (alpha=0.05, df={df_resid1}) = +-{t_crit:.4f}")
```

$t = 5.3265$, with $df = 67$. Since $|t| = 5.3265 > t_{crit} = 1.9960$ (p < 0.0001), we reject $H_0$. There is statistically significant evidence that $\beta \neq 0 - X_1$ has a significant linear relationship with the response, holding $X_2$, $X_3$, $X_4$ constant. 

#### Question 4

**Compute the R-Squared value for Model 1, using information from the ANOVA table. Interpret this statistic.**
$$R^2 = \frac{SS_{Model}}{SS_{Total}} = 1 - (\frac{SS_{Residual}}{SS_{Total}})$$

```python
SS_model1 = model1['anova']['Model']['ss']
SS_resid1 = model1['anova']['Residuals']['ss']
SS_total = model1['anova']['Total']['ss']

R2_1 = SS_model1 / SS_total
R2_1_check = 1 - SS_resid1/SS_total

print(f"R^2 = {SS_model1} / {SS_total} = {R2_1:.4f}")
print(f"check (1 - SSE/SST) = {R2_1_check:.4f}")
print(f"Reported in R output: {model1['mult_r2']}")
```

$R^2 = 0.7713$. Approximately 77.13% of the variability in the response is explained by the four predictors ($X_1 - X_4$) jointly in Model 1. 

#### Question 5

**Compute the Adjusted R-Squared value for Model 1. Discuss why Adjusted R-squared and the R-squared value are different.**
$$R_{adj}^2 = 1 - (1 - R^2) \frac{n-1}{n-k-1}$$

```python
k1 = model1['n_predictors']
adjR2_1 = 1 - (1 - R2_1) * (n - 1) / (n - k1 - 1)

print(f"n = {n}, k = {k1}")
print(f"Adjusted R^2 = 1 - (1-{R2_1:.4f}) * ({n}-1)/({n}-{k1}-1) = {adjR2_1:.4f}")
print(f"Reported in R output: {model1['adj_r2']}")
```

Adjusted $R^2 = 0.7577$, slightly lower than $R^2 = 0.7713$. Ordinary $R^2$ increases every time an additional predictor is added to the model, regardless of whether that predictor is actually useful. Adjusted $R^2$ imposes a penalty for the number of predictors relative to the sample size, so it only increases if a new predictor actually improves the fit by more than what would be expected by chance. This makes Adjusted $R^2$ more appropriate for comparing models with different numbers of predictors. 

#### Question 6

**Write out the null and alternate hypothesis for the Overall F-test.**

$$H_0 = \beta_1 = \beta_2 = \beta_3 = \beta_4 = 0$$
$$H_a = \text{at least one } \beta_j \neq 0, j = 1,2,3,4$$

#### Question 7

**Compute the F-statistic for the Overall F-test. Conduct the hypothesis test and interpret the result.**

$$F = \frac{MS_{Model}}{MS_{Residual}} = \frac{SS_{Model}/df_{Model}}{SS_{Residual}/df_{Residual}}$$

```python
MS_model1 = model1['anova']['Model']['ms']
MS_resid1 = model1['anova']['Residuals']['ms']
df_model1 = model1['anova']['Model']['df']
df_resid1 = model1['anova']['Residuals']['df']

F1 = MS_model1 / MS_resid1
p_F1 = 1 - stats.f.cdf(F1, df_model1, df_resid1)
F1_crit = stats.f.ppf(0.95, df_model1, df_resid1)

print(f"F = {MS_model1} / {MS_resid1} = {F1:.4f}")
print(f"df = ({df_model1}, {df_resid1})")
print(f"p-value = {p_F1:.6f}")
print(f"F critical (alpha=0.05) = {F1_crit:.4f}")
```

$F = 56.4825$ with $df = (4,67)$. Since $F = 56.4825 > F_{crit} = 2.5087$ (p < 0.0001), we reject $H_0$. The overall regression is statistically significant. The four combined predictors explain a significant amount of the variability in the response.

### Model 1 vs. Model 2

#### Question 8

**Does Model 1 nest Model 2 or does Model 2 nest Model 1? Explain.**
- Model 1: response $~ X_1 + X_2 + X_3 + X_4$
- Model 2: response $~ X_1 + X_2 + X_3 + X_4 + X_5 + X_6$

Model 1's predictors are a subset of Model 2's predictors. Model 1 can be obtained from Model 2 by restricting $\beta_5 = \beta_6 = 0$. Model 2 nests Model 1. 

#### Question 9

**Write out the null and alternate hypotheses for nested F-test using Model 1 and Model 2.**
$$H_0 = \beta_5 = \beta_6 = 0$$
$$H_a = \text{at least one of } \beta_5, \beta_6 \neq 0$$

#### Question 10

**Compute the F-statistic for a nested F-test using Model 1 and Model 2. Conduct the hypothesis test and interpret the results.**
$$F = \frac{(SSE{reduced} - SSE{full})/(df_{reduced} - df_{full})}{SSE_{full}/df_{full}}$$

where "reduced" = Model 1 and "full" = Model 2

```python
SSE_reduced = model1['anova']['Residuals']['ss']
df_reduced = model1['anova']['Residuals']['df']

SSE_full = model2['anova']['Residuals']['ss']
df_full = model2['anova']['Residuals']['df']

df_num = df_reduced - df_full
F_nested = ((SSE_reduced - SSE_full) / df_num) / (SSE_full / df_full)
p_nested = 1 - stats.f.cdf(F_nested, df_num, df_full)
F_nested_crit = stats.f.ppf(0.95, df_num, df_full)

print(f"SSE_reduced (Model 1) = {SSE_reduced}, df = {df_reduced}")
print(f"SSE_full (Model 2) = {SSE_full}, df = {df_full}")
print(f"numerator df = {df_reduced} - {df_full} = {df_num}")
print()
print(f"F = [({SSE_reduced} - {SSE_full}) / {df_num}] / [{SSE_full} / {df_full}]")
print(f"F = {F_nested:.4f}")
print(f"df = ({df_num}, {df_full})")
print(f"p-value = {p_nested:.4f}")
print(f"F critical (alpha=0.05) = {F_nested_crit:.4f}")
```

$F = 3.2778$ with $df = (2,65), p = 0.0440$. Since $F = 3.2778 > F_{crit} = 3.1381$, we reject $H_0$. Adding $X_5$ and $X_6$ to Model 1 produces a statistically significant improvement in fit at the 95% confidence level, so Model 2 is preferred over Model 1.
